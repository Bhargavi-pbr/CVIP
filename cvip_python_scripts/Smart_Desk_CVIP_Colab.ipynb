```python
# Step 2: Import Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
import math
from ultralytics import YOLO

print('All libraries successfully imported!')
```

```python
# Step 3: Classical CVIP Pipeline Visualization
# Create synthetic test desk scene
img = np.ones((480, 640, 3), dtype=np.uint8) * 45
cv2.rectangle(img, (80, 60), (560, 420), (30, 25, 20), -1)
cv2.rectangle(img, (200, 120), (440, 320), (120, 120, 130), -1)
cv2.circle(img, (140, 200), 28, (30, 80, 200), -1)
cv2.rectangle(img, (460, 160), (540, 300), (40, 160, 60), -1)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (5, 5), 1.4)
edges = cv2.Canny(blurred, 40, 110)
_, otsu = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); axes[0].set_title('Original Desk')
axes[1].imshow(gray, cmap='gray'); axes[1].set_title('Grayscale')
axes[2].imshow(edges, cmap='gray'); axes[2].set_title('Canny Edges')
axes[3].imshow(otsu, cmap='gray'); axes[3].set_title('Otsu Binarization')
for ax in axes: ax.axis('off')
plt.show()
```

```python
# Step 4: YOLOv8 Object Detection & Spill Hazard Calculation
model = YOLO('yolov8n.pt')

# Run inference on the scene
results = model(img)[0]
print(f'Detected {len(results.boxes)} objects!')

# Display annotated output
annotated_frame = results.plot()
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('YOLOv8 Detection Output')
plt.show()
```
